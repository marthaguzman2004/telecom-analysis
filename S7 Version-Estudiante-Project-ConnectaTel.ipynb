## 🧩Paso 3: Limpieza básica de datos

### 3.1 Corregir sentinels y fechas imposibles
**🎯 Objetivo:**  
Aplicar reglas de limpieza para reemplazar valores sentinels y corregir fechas imposibles.

**Instrucciones:**  
- En `age`, reemplaza el sentinel **-999** con la mediana.
- En `city`, reemplaza el sentinel `"?"` por valores nulos (`pd.NA`).  
- Marca como nulas (`pd.NA`) las fechas fuera de rango.

```python
# Reemplazar -999 por la mediana de age
age_mediana = users['age'].median()
users['age'] = users['age'].replace(-999, pd.NA).fillna(age_mediana)

# Verificar cambios
users['age'].describe()
```

```python
# Reemplazar ? por NA en city

users["city"] = users["city"].replace(["?"], pd.NA)

# Verificar cambios

users["city"].describe()
```

```python
# Marcar fechas futuras como NA para reg_date

fechas_futuras = users["reg_date"].dt.year == 2026
users.loc[fechas_futuras, "reg_date"] = pd.NaT

# Verificar cambios
users["reg_date"].describe()
```

## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico. 

**Instrucciones:**: 
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python
# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int) #conocer el total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int) #conocer el total de llamadas

# Agrupar información por usuario

usage_agg = usage.groupby("user_id")[["is_text", "is_call", "duration"]]\
        .agg({"is_text": "sum", "is_call": "sum", "duration": "sum"})\
        .reset_index()


# observar resultado
usage_agg.head(3)
```

```python
# Renombrar columnas
usage_agg = usage_agg.rename(columns={"is_text": "cant_mensajes", "is_call": "cant_llamadas", "duration": "cant_minutos_llamadas"})


# observar resultado
usage_agg.head(3)
```

```python
# Combinar la tabla agregada con el dataset de usuarios
user_profile = pd.merge(users, usage_agg, on="user_id", how="left")
user_profile.head(5)


```

```python
# Visualizando usando BoxPlot 
columnas_numericas = ['age', 'cant_mensajes', 'cant_llamadas', 'cant_minutos_llamadas']

for col in columnas_numericas:
    plt.figure(figsize=(9,3))
    sns.boxplot(x=user_profile[col])
    plt.title(f'Boxplot: {col}')
    plt.show()
```

💡Insights: 

- Age: no presenta outliers.
- cant_mensajes: presenta outliers hacia la derecha que representan pocos usuarios con arriba de 12 mensajes.
- cant_llamadas: presenta outliers hacia la derecha que representan pocos usuarios con arriba de 11 llamadas.
- cant_minutos_llamadas: presenta outliers hacia la derecha que representan algunos usuarios con llamadas arriba de 60 minutos.


```python
# Calcular límites con el método IQR

columnas_limites = ['cant_mensajes', 'cant_llamadas', 'cant_minutos_llamadas']
limites_superiores = {}

for col in columnas_limites:
    q1 = user_profile[col].quantile(0.25)
    q3 = user_profile[col].quantile(0.75)
    IQR = q3 - q1
    upper = q3 + 1.5 * IQR
    limites_superiores[col] = upper

limites_superiores

    


```

```python
# Revisa los limites superiores y el max, para tomar la decisión de mantener los outliers o no
user_profile[columnas_limites].describe()
```


💡Insights: 
- cant_mensajes: Mantenerlos porque representan usuarios reales con un uso más alto de mensajes, útil para segmentarlos como clientes de alto consumo, no son errores de datos.
- cant_llamadas: Mantenerlos porque representan usuarios reales con un uso más alto de llamadas, útil para segmentarlos como clientes de alto consumo, no son errores de datos
- cant_minutos_llamadas: Mantenerlos porque representan usuarios reales con llamadas mas largas, útil para segmentarlos como clientes de alto consumo, no son errores de datos

## 🧩Paso 6: Segmentación de Clientes

### 6.1 Segmentación de Clientes Por Uso

🎯 **Objetivo:** Clasificar a cada usuario en un grupo de uso (Bajo uso, Uso medio, Alto uso) basándose en la cantidad de llamadas y mensajes registrados.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_uso` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python
# Crear columna grupo_uso



def clasificar_uso(row):
    llamadas = row["cant_llamadas"]
    mensajes = row["cant_mensajes"]


    if llamadas <5 and mensajes < 5:
        return 'Bajo uso'
    elif llamadas <10 and mensajes <10:
        return 'Uso medio'
    else:
        return 'Alto uso'

user_profile["grupo_uso"] = user_profile.apply(clasificar_uso, axis=1)


    
```

```python
# verificar cambios
user_profile.head()
```

### 6.2 Segmentación de Clientes Por Edad

🎯 **Objetivo:**: Clasificar a cada usuario en un grupo por **edad**.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_edad` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones y asigna:
  - `'Joven'` cuando age < 30
  - `'Adulto'` cuando age < 60
  - `'Adulto Mayor'` para el resto de casos

```python
# Crear columna grupo_edad

def clasificar_edad(row):
    edad = row["age"]


    if edad < 30:
        return 'Joven'
    elif edad <60:
        return 'Adulto'
    else:
        return 'Adulto Mayor'

user_profile["grupo_edad"] = user_profile.apply(clasificar_edad, axis=1)


    
```

- El alto porcentaje de valores nulos en churn_date (88.35%) no representa un error de datos, sino que indica que la gran mayoría de los clientes siguen activos (no han dado de baja el servicio).


🔍 **Segmentos por Edad**
- La mayoría de los clientes son adultos, después le siguen adultos mayores y el grupo mas bajo es el de jovenes.


📊 **Segmentos por Nivel de Uso**
- La mayoría de los clientes estan en uso medio, después siguen los de bajo uso y al final los de alto uso.


**Segmentos valiosos para ConnectaTel**
El segmento de "uso medio" y el de "adultos" son los más valiosos porque concentran la mayor cantidad de clientes.  El segmento de "alto uso" es valioso por el ingreso individual alto que representa aunque sea un grupo pequeño.  El segmente de "jóvenes" es valioso por el potencial de crecimiento/innovación a futuro.


**Outliers y su implicación para el negocio**
Se encontraron outliers en las variables "cant_mensajes", "cant_llamadas" y "cant_minutos_llamadas".   La implicación sería una oportunidad para ofrecer planes premium/con mayor cantidad de minutos, mensajes o datos a estos clientes de alto consumo (upsell).
Pero también implica un riesgo si no están en el plan adecuado ya que la empresa puede estar perdiendo ingresos por consumo no facturado correctamente, o generando cobros inesperados que afecten la satisfacción del cliente.


💡 **Recomendaciones**
- Se recomienda hacer campañas de retención para adultos ya que es el grupo mayoritario y perderlos tendría un mayor impacto.  Los jóvenes son un buen target para nuevos servicios o funcionalidades, porque suelen ser mas receptivos a innovaciones. 
En cuestión de uso se pudiera ofrecer un plan con mas minutos/mensajes al bloque de alto uso, al bloque de bajo uso ofrecerle un plan mas económico y el bloque de uso medio es ideal para campañas de fidelizacion o upselling moderado.
